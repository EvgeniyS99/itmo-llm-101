# Обязательные задачи

### 1. Поменять логику `NextHintSubgraph`
Теперь хочется, чтобы агент следовал следующей логике:
   * Пусть теперь агент выдает не просто подсказку и вопрос "Понятна ли тебе подсказка", а подсказку с наводящим вопросом
   * Пусть теперь студент дает фидбек не в виде "да/нет", а пишет ответ на поставленный тьютором вопрос
   * Если студент отвечает верно, агент его переводит к следующей подсказке-вопросу, если нет -- продолжает работать с текущей подсказкой (например, даёт дополнительное объяснение или новые уточняющий вопрос)

### 2. Улучшить ветку с `full_solution_node`
Когда агент отправляет решение сразу -- в этом случае у студента могут возникнуть вопросы по решению. Тогда хочется следующее:
* Тьютор задает вопрос по типу "Есть ли вопросы по решению?". Если у студента есть вопрос, он его пишет, и в этом случае вызывается `Explainer` нода
* Это может быть та же `Explainer` нода, которая уже используется в `NextHintSubgraph`, либо новая
   
### 3. Дописать метод `_stream_from_input` класса `BaseGraph`
* Сейчас этот метод работает корректно только когда `mode == 'updates'`. Причем, обратите внимание, он стримит в чат только сообщения нод, определенных в `self.show_nodes`
* Но хочется также, чтобы агент работал и с `mode == 'messages'`. Это даст возможность стримить сообщения от LLM чанками в случае, когда вам необходимо **on the fly** получать сообщения от LLM (это может быть полезно, например, в `Explainer` ноде)

### 4. Запуск модели
В репозитории, в папке `tgi`, приведён пример запуска LLM на базе Text Generation Inference (TGI) с использованием GPU.
Вы можете использовать этот пример как основу для запуска модели также через Ollama или vLLM, изменив файл `docker-compose.yaml`.

Также допускается запуск модели без Docker, напрямую из окружения, например с помощью команд:
`!vllm serve` ... или `!ollama serve ....`

Вариант с использованием **Docker-контейнера является предпочтительным**.

### 5. Реализация альтернативного способа запуска LLM
Сейчас приложение работает с моделью, поднятой на OpenAI-совместимом сервере.\
Для это есть функция `_make_chatopenai_llm_from_config` в файле `models.py`.
* Вам нужно реализовать свою функцию (например, `_make_ollama_llm_from_config`) в зависимости от способа, которым вы сервете свою модель
* Возможно, вам придется добавить новые параметры в конфиг модели (`ModelSettings` в файле `settings.py`)
* При желании можно поднимать модель в одном Docker-контейнере с приложением (в этом случае потребуется небольшая доработка `docker-compose.yaml`)

### 6. При необходимости изменить tool
В текущей реализации tool `math_solver` предполагает использование отдельной специализированной математической LLM. Если вы планируете использовать одну и ту же LLM для всего приложения (без выделенной math-модели), то вам следует адаптировать tool под выбранную модель: добавить `structured_output`или поменять `system_prompt` (вспомните реализованный вами tool в домашнем задании по LangChain (**hw_2**)).

#  Как сдавать дз
## Опция номер 1
1. В репозитории создаете ветку со своим именем и ведете разработку в ней. **Пожалуйста, не делайте коммитов в ветку `main`**.
2. Ваше финальное приложение может быть завернуто в докер контейнер, либо это может быть только файл `run_app`, который запускается через `python -m run_app` (лучше первое)
3. Как только вы получаете вашу фильную версию -- сообщаете об этом

## Опция номер 2
* Подойти на семинар и показать мне работающее приложение
* Также вы можете показать предыдущее дз по реализацию простого чат бота
* Если оба приложения удовлетворяют критериям (см. ниже), вы получаете сразу оценку (при условии что предыдущие два дз также выполнены)

## Критерии проверки:
* Приложение запускается
* Все обязательные задачи реализованы и работают
* Вся предыдущая логика приложения остаётся корректной
* Код не вырвиглазный :)
